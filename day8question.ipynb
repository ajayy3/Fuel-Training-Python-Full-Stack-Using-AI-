```python
#4 find outliers in a list of salaries using the IQR rule

import pandas as pd

salaries = [30000, 35000, 40000, 42000, 45000, 50000, 52000, 55000, 60000, 200000]

q1 = pd.Series(salaries).quantile(0.25)
q3 = pd.Series(salaries).quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = [salary for salary in salaries
            if salary < lower_bound or salary > upper_bound]

print("Outliers:", outliers)
```
