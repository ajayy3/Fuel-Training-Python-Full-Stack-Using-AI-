```python
#Dataset
import pandas as pd
import numpy as np

raw = pd.DataFrame({
    "name":["riya","aman","NEHA","karan","aman","isha","rohan"],
    "age":[21,22,np.nan,23,22,20,250],
    "city":["pune","mumbai","pune","none","mumbai","pune","none"],
    "marks":["85","62","91","45","62","78","70"],
    "grade":["A","B","A","C","B","B","B"]
})
print(raw)
```

```python
#dduplicates and missing values

print("Duplicates:",df.duplicated().sum())
print(df[df.duplicated(keep=False)]),
```

```python
#Removing duplicates

df=df.drop_duplicates().reset_index(drop=True)
print(df)
```

```python
#Finding missing values

print(df.isnull().sum())
```

```python

# Calculate Q1 and Q3
q1 = df["Age"].quantile(0.25)
q3 = df["Age"].quantile(0.75)

# Calculate IQR
iqr = q3 - q1

# Calculate outlier limits
low = q1 - 1.5 * iqr
high = q3 + 1.5 * iqr

print("Q1, Q3, IQR:", q1, q3, iqr)
print("Limits:", low, high)

# Display outlier rows
print(df[(df["Age"] < low) | (df["Age"] > high)])
```

```python

import pandas as pd
import numpy as np

# Convert age column to numeric values
df["age"] = pd.to_numeric(df["age"], errors="coerce")

# Replace ages greater than 100 with missing values
df.loc[df["age"] > 100, "age"] = np.nan

# Fill missing ages with the median
df["age"] = df["age"].fillna(df["age"].median())

# Fill missing cities with "unknown"
df["city"] = df["city"].fillna("unknown")

# Display cleaned data
print(df)

# Count remaining missing values
print("Missing:", df.isnull().sum().sum())
```

```python
#labelencoder

from sklearn.preprocessing import LabelEncoder

le=LabelEncoder()
print(le.fit_transform(df["grade"]).tolist())
print(le.classes_.tolist())
```
