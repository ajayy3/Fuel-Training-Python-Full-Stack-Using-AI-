```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
#plot square of numbers from 1 to 10 as a line charts with market,titleand labels

a = np.arange(1,11)
b = a**2 
plt.plot(a, b)
plt.xlabel("Numbers")
plt.ylabel("Squares")
plt.title("Square of Numbers")
plt.show()

```

```python
#draw a bar chart of cities and thier population.use a different colour
cities = ["New York", "Los Angeles", "Chicago", "Houston", "Phoenix"]
population = [8336817, 3979575, 2746385, 2320268, 1680992]
plt.bar(cities, population, color=['red', 'blue', 'green', 'yellow', 'purple'])
plt.xlabel("Cities")
plt.ylabel("Population")
plt.title("Population of Cities")
plt.show()
```

```python
#draw a pie chart of how you spend a 24 hour day
activities = ["Work", "Sleep", "Eat", "Exercise", "Leisure"]
time_spent = [8, 8, 2, 1, 5]
plt.pie(time_spent, labels=activities, autopct='%1.1f%%')
plt.title("How I Spend My Day")
plt.show()
```

```python
#plot study hours against marks as a scatter plot for 10 student
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6, 7, 2, 4, 8]
marks =       [45, 50, 55, 60, 65, 72, 78, 52, 62, 85]

plt.scatter(study_hours, marks)
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs. Marks for 10 Students")
plt.show()
```

```python
#draw two lines (python and java scores) in one chart with a legend 
import matplotlib.pyplot as plt

students = ["A", "B", "C", "D", "E"]
python_scores = [85, 90, 78, 92, 88]
java_scores = [80, 85, 82, 89, 91]

plt.plot(students, python_scores, marker="o", label="Python")
plt.plot(students, java_scores, marker="o", label="Java")
plt.xlabel("Students")
plt.ylabel("Scores")
plt.title("Python and Java Scores")
plt.legend()
plt.show()
```

```python
#create a figure with 2  rows and 2 colums of subplots showing 4 different charts
import matplotlib.pyplot as plt

fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# Line chart
axs[0, 0].plot([1, 2, 3, 4], [10, 15, 12, 20], marker="o")
axs[0, 0].set_title("Line Chart")
axs[0, 0].set_xlabel("X")
axs[0, 0].set_ylabel("Value")

# Bar chart
axs[0, 1].bar(["A", "B", "C"], [5, 8, 6], color="orange")
axs[0, 1].set_title("Bar Chart")

# Scatter plot
axs[1, 0].scatter([1, 2, 3, 4], [4, 7, 5, 9], color="green")
axs[1, 0].set_title("Scatter Plot")
axs[1, 0].set_xlabel("X")
axs[1, 0].set_ylabel("Y")

# Pie chart
axs[1, 1].pie([40, 30, 20, 10], labels=["A", "B", "C", "D"], autopct="%1.1f%%")
axs[1, 1].set_title("Pie Chart")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'City': ['New York', 'San Francisco', 'Los Angeles', 'Chicago'],
    'Marks': [90, 85, 78, 92],
    'Attendance': [90, 75, 98, 40]
})
grouped_data = df.groupby('City')['Marks'].mean()
grouped_data.plot(kind='bar', color='skyblue')
plt.title('Average Marks by City')
plt.xlabel('City')
plt.ylabel('Average Marks')
plt.show()
```

```python
#create a dashboard for 6 month of shop data
#line chart : monthly revenue 
#bar chart : orders per month 
#print the best month and total revenue 
#save the dashboard as dashboard.png 
#bonus : add a third chart showing average revenue per order

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data = pd.DataFrame({
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Revenue': [1000,2000,3000,4000,5000,6000],
    'Orders': [100, 120, 90, 150, 110, 130]
})
monthly_revenue = data.set_index('Month')['Revenue']
monthly_orders = data.set_index('Month')['Orders']
best_month = monthly_revenue.idxmax()
total_revenue = monthly_revenue.sum()
print(f"Best month: {best_month}")
print(f"Total revenue: ${total_revenue}")
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
monthly_revenue.plot(kind='line', marker='o', color='blue', linewidth=2)
plt.title('Monthly Revenue')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.grid(True)
plt.subplot(1, 2, 2)
monthly_orders.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Orders per Month')
plt.xlabel('Month')
plt.ylabel('Orders')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.savefig('dashboard.png')
plt.show()
```
