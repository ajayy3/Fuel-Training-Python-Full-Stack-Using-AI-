```python
import matplotlib.pyplot as plt

day = [1, 2, 3, 4, 5]

a = [10, 20, 15, 25, 30]
b = [8, 14, 18, 20, 26]

plt.plot(day, a, color="blue", marker="o",
         linestyle="--", label="Python")

plt.plot(day, b, color="red", marker="s",
         linestyle="-", label="Java")

plt.title("Weekly Scores")
plt.xlabel("Days")
plt.ylabel("Scores")
plt.legend()
plt.grid(True)
plt.show()
```

```python
subjects = ["Math", "Science", "English", "History"]
marks = [85, 90, 78, 92]

plt.bar(subjects, marks, color=("yellow"))
plt.title("Marks in Subjects")
plt.xlabel("marks")
plt.show()
```

```python
import numpy as np
```

```python
np.random.seed(0)
data = np.random.normal(60, 5, 70)

plt.hist(data, bins=10, color="skyblue", edgecolor="black")
plt.title("Distribution of marks")
plt.xlabel("Marks")
plt.ylabel("Students")
plt.show()
```

```python
hours = [1, 2, 3, 4, 5,6,7,8]
scores = [10, 12, 15, 25, 30,35,40,45]
plt.scatter(hours, scores, color="green",marker="o")
plt.title("Hours vs Scores")
plt.xlabel("Hours")
plt.ylabel("Scores")
plt.show()
```

```python
sizes = [20, 50, 80, 200, 500]
names = ["A", "B", "C", "D", "E"]
plt.pie(sizes, labels=names, autopct="%1.1f%%")
plt.title("Pie Chart Example")
plt.show()
```

```python
#figsize

days = [1, 2, 3, 4, 5, 6, 7]
temp = [30, 32, 31, 29, 28, 27, 26]
plt.figure(figsize=(10, 4))
plt.plot(days,temp,marker="o",linestyle="--",color="skyblue")
plt.title("Temperature over Days")
plt.show()
```

```python
subjects = ["Math", "Science", "English", "History"]
marks = [85, 90, 78, 92]

fig,ax=plt.subplots(1,2,figsize=(10,4))
ax[0].plot(days,temp)
ax[0].set_title("Temperature over Days")
ax[1].bar(subjects, marks)
ax[1].set_title("bars")
plt.tight_layout()
plt.show()
```

```python
#saving a chart

plt.plot(days,temp)
plt.savefig("temp.png",dpi=150)
plt.show()
```

```python
#  Quick Reference - Matplotlib
# Function                    Purpose
# -----------------------------------------------
# plt.plot()                  Line chart
# plt.bar() / plt.barh()      Bar charts
# plt.hist()                  Histogram
# plt.scatter()               Scatter plot
# plt.pie()                   Pie chart
# plt.title()                 Set chart title
# plt.xlabel()                Set X-axis label
# plt.ylabel()                Set Y-axis label
# plt.legend()                Display legend
# plt.grid()                  Display grid lines
```
