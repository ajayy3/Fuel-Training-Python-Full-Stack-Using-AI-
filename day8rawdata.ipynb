```python

# Calculate Q1 and Q3
q1 = df["Age"].quantile(0.25)
q3 = df["Age"].quantile(0.75)

# Calculate IQR
iqr = q3 - q1

# Calculate outlier limits
low = q1 - 1.5 * iqr
high = q3 + 1.5 * iqr

print("Q1, Q3, IQR:", q1, q3, iqr)
print("Limits:", low, high)

# Display outlier rows
print(df[(df["Age"] < low) | (df["Age"] > high)])
```

```python
import numpy as np
df.loc[df["Age"]>100,"Age"]=np.nan

df["Age"]=df["Age"].fillna(df["Age"].median)
df["Gender"]=df["Gender"].fillna("unknown")
print(df)
print("missing:",df.isnull().sum().sum())
```
